```python
import os

# Function to extract characters from transcriptions
def extract_chars_from_file(file_path):
    charset = set()
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            for line in f:
                line = line.strip()
                if line:
                    # Split by spaces and take everything after the filename (first element)
                    parts = line.split(' ')
                    if len(parts) > 1:
                        # Join all parts after the filename to get the transcription
                        transcription = ' '.join(parts[1:])
                        # Add all characters to charset
                        charset.update(transcription)
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
    return charset

# List of source files
source_files = [
    'nan_raw_biner/train/lines.txt',
    'nan_raw_biner/test/lines.txt', 
    'nan_raw_biner/validation/lines.txt'
]

# Special tokens to include first
special_tokens = ['<UNK>', '<SPACE>', '<PAD>', '|']

# Collect all unique characters
all_chars = set()
all_chars.update(special_tokens)

# Extract characters from each file
for file_path in source_files:
    if os.path.exists(file_path):
        chars = extract_chars_from_file(file_path)
        all_chars.update(chars)
        print(f"Extracted {len(chars)} characters from {file_path}")
    else:
        print(f"File not found: {file_path}")

# Convert to sorted list
char_list = sorted(list(all_chars))

# Write to CHAR_LIST file
output_path = 'Sets/CHAR_LIST_2'
os.makedirs(os.path.dirname(output_path), exist_ok=True)

with open(output_path, 'w', encoding='utf-8') as f:
    for char in char_list:
        f.write(char + '\n')

print(f"\nGenerated CHAR_LIST with {len(char_list)} characters:")
print("First 10 characters:", char_list[:10])
print("Last 10 characters:", char_list[-10:])
print(f"\nCHAR_LIST saved to: {output_path}")
```
